```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Simulate blood sugar data (mg/dL)
np.random.seed(42)
normal_group = np.random.normal(100, 10, 140)   # Normal patients
diabetic_group = np.random.normal(180, 15, 60)  # Diabetic patients
blood_sugar = np.concatenate([normal_group, diabetic_group])

# Create KDE plot
plt.figure(figsize=(8, 5))
sns.kdeplot(blood_sugar, fill=True, color='pink', linewidth=2)

# Labels and title
plt.title('Density Plot of Blood Sugar Levels (n=200 Patients)', fontsize=14)
plt.xlabel('Blood Sugar Level (mg/dL)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(alpha=0.3)
plt.show()
```
